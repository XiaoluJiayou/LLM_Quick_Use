# UniT2IXL 出图示例
需要单独安装模型官方实现（`pipeline_unit2ixl` 未包含在本仓库）。设置 `DATA_CSV`、`UNIT2IXL_MODEL_DIR`、`OUTPUT_DIR`；处理实际 CSV 全部行。每次运行可能覆盖同名示例图片，正式可续跑评测请使用 t2i_qwen_singguard。


```python
import pandas as pd
from pipeline_unit2ixl import UniT2IXLPipeline
import os
from pathlib import Path

from llm_quick_use.records import image_prompt_records
```

```python
data = pd.read_csv(os.environ.get("DATA_CSV", "jade_benchmark_zh.csv"), dtype={"ID": str, "提示词": str})
records = image_prompt_records(data.to_dict(orient="records"), base_seed=0)
print(f"待生成 {len(records)} 张图片")
```

```python
data.head()
len(data)
```

```python
for record in records[:10]:
    print(record["prompt"])
```

```python
pipeline = UniT2IXLPipeline.from_pretrained(os.environ.get("UNIT2IXL_MODEL_DIR", "/root/autodl-tmp/UnicomAI/UniT2IXL"))
# pipeline.load_lora_weights("lora.safetensors")
pipeline = pipeline.to("cuda:0")
```

```python
base_path = Path(os.environ.get("OUTPUT_DIR", "img_result"))
base_path.mkdir(parents=True, exist_ok=True)
for record in records:
    image = pipeline(prompt=record["prompt"], guidance_scale=7.5, target_size=(1024, 1024)).images[0]
    image.save(base_path / f"{record['id']}.png")
```
